# 2 — Autoencoder (VAE): ricostruzione

Primo stadio: un **autoencoder** (`AutoencoderKlMaisi`) comprime i volumi `256^3` in latenti `4x64x64x64` (compressione 4x per asse) e li ricostruisce.

La qualita' di ricostruzione definisce il **tetto teorico** della generazione: l'LDM lavora nel latente, quindi non puo' produrre dettagli che il decoder non ricostruisce.

**VAE adottato.** L'iterazione corrente ha sperimentato un warmup lineare del learning rate per il VAE, che ha prodotto ricostruzioni **peggiori** (SSIM ~0.85) rispetto allo schema a gradini precedente. La pipeline riusa quindi il VAE precedente, migliore, sia per l'encoding sia per il decoding.

**Metriche del VAE adottato (test set)**: SSIM ~0.986, PSNR ~32.9 dB, L1 ~0.006.

```python
import os, sys, json, glob
ROOT=os.path.abspath("..")
if ROOT not in sys.path:
    sys.path.insert(0, ROOT)
os.chdir(ROOT)

import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
%matplotlib inline

def show_ortho(vol, title=None, ax_row=None, cmap="gray"):
    """Mostra le 3 viste ortogonali (centrali) di un volume 3D."""
    if hasattr(vol, "numpy"):
        vol=vol.numpy()
    vol=np.asarray(vol).squeeze()
    x, y, z=[s // 2 for s in vol.shape]
    views=[(np.rot90(vol[x, :, :]), "Sagittale"),
             (np.rot90(vol[:, y, :]), "Coronale"),
             (np.rot90(vol[:, :, z]), "Assiale")]
    axes=ax_row if ax_row is not None else plt.subplots(1, 3, figsize=(12, 4))[1]
    for a, (img, name) in zip(axes, views):
        a.imshow(img, cmap=cmap); a.set_title(name); a.axis("off")
    if title and ax_row is None:
        plt.suptitle(title); plt.tight_layout()
    return axes
```

**Osservazione.** Le ricostruzioni preservano le strutture anatomiche con un leggero *smoothing* delle alte frequenze (texture corticale, foliazioni cerebellari), mantenendo pero' bordi netti fra sostanza grigia e bianca — coerente con SSIM ~0.986.

**Il VAE non e' il collo di bottiglia.** Il confronto fra queste ricostruzioni (nitide) e i campioni *generati* prima dell'introduzione dell'autoguidance (nebbiosi, cfr. notebook 3) mostra che il divario di nitidezza non nasce dalla compressione: se il limite fosse il VAE, le generazioni assomiglierebbero alle ricostruzioni. Il collo di bottiglia risiede nella capacita' dell'LDM di modellare la distribuzione dei latenti, ed e' li' che sono stati concentrati gli interventi.
